### Task 1: Stacked Chart for Composition of Job Satisfaction Across Age Groups


##### 1. Stacked Chart of Median `JobSatPoints_6` and `JobSatPoints_7` for Different Age Groups


Visualize the composition of job satisfaction scores (`JobSatPoints_6` and `JobSatPoints_7`) across various age groups. This will help in understanding the breakdown of satisfaction levels across different demographics.



```python
data = df[['Age', 'JobSatPoints_6', 'JobSatPoints_7']].copy()

data['JobSatPoints_6'] = pd.to_numeric(data['JobSatPoints_6'], errors='coerce')
data['JobSatPoints_7'] = pd.to_numeric(data['JobSatPoints_7'], errors='coerce')

age_order = [
    'Under 18 years old',
    '18-24 years old',
    '25-34 years old',
    '35-44 years old',
    '45-54 years old',
    '55-64 years old',
    '65 years or older'
]

result = data.groupby('Age')[['JobSatPoints_6', 'JobSatPoints_7']].median()
result = result.reindex(age_order).dropna(how='all')

result.plot(kind='bar', stacked=True, figsize=(10,6))

plt.title('Median Job Satisfaction by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Median Job Satisfaction Score')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Job Satisfaction')
plt.tight_layout()
plt.show()
```

##### Stacked Chart of `JobSatPoints_6` and `JobSatPoints_7` for Employment Status


Create a stacked chart to compare job satisfaction (`JobSatPoints_6` and `JobSatPoints_7`) across different employment statuses. This will show how satisfaction varies by employment type.


```python
data = df[['Employment', 'JobSatPoints_6', 'JobSatPoints_7']].dropna(subset=['Employment']).copy()

data['Employment'] = data['Employment'].str.split(';')
data = data.explode('Employment')
data['Employment'] = data['Employment'].str.strip()

data['JobSatPoints_6'] = pd.to_numeric(data['JobSatPoints_6'], errors='coerce')
data['JobSatPoints_7'] = pd.to_numeric(data['JobSatPoints_7'], errors='coerce')

top_employment = data['Employment'].value_counts().head(6).index
data = data[data['Employment'].isin(top_employment)]

result = data.groupby('Employment')[['JobSatPoints_6', 'JobSatPoints_7']].median()

result.plot(kind='bar', stacked=True, figsize=(11,6))

plt.title('Job Satisfaction by Employment Status')
plt.xlabel('Employment Status')
plt.ylabel('Median Job Satisfaction Score')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Job Satisfaction')
plt.tight_layout()
plt.show()
```

### Task 2: Stacked Chart for Compensation and Job Satisfaction by Age Group


##### This stacked chart visualizes the composition of compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`) specifically for respondents aged 30-35.


```python
data = df[
    df['Age'].isin(['25-34 years old', '35-44 years old'])
][['Age', 'ConvertedCompYearly', 'JobSatPoints_6']].copy()

data['ConvertedCompYearly'] = pd.to_numeric(data['ConvertedCompYearly'], errors='coerce')
data['JobSatPoints_6'] = pd.to_numeric(data['JobSatPoints_6'], errors='coerce')

result = data.groupby('Age')[['ConvertedCompYearly', 'JobSatPoints_6']].median()

result_normalized = result.div(result.max())

result_normalized.plot(kind='bar', stacked=True, figsize=(8,6))

plt.title('Compensation and Job Satisfaction for Relevant Age Groups')
plt.xlabel('Age Group')
plt.ylabel('Normalized Median')
plt.xticks(rotation=0)
plt.legend(title='Metric')
plt.tight_layout()
plt.show()
```

##### Stacked Chart of Median Compensation and Job Satisfaction Across Age Group


Compare the median compensation and job satisfaction metrics across different age groups. This helps visualize how compensation and satisfaction levels differ by age.


```python
data = df[['Age', 'ConvertedCompYearly', 'JobSatPoints_6']].copy()

data['ConvertedCompYearly'] = pd.to_numeric(data['ConvertedCompYearly'], errors='coerce')
data['JobSatPoints_6'] = pd.to_numeric(data['JobSatPoints_6'], errors='coerce')

age_order = [
    'Under 18 years old',
    '18-24 years old',
    '25-34 years old',
    '35-44 years old',
    '45-54 years old',
    '55-64 years old',
    '65 years or older'
]

result = data.groupby('Age')[['ConvertedCompYearly', 'JobSatPoints_6']].median()
result = result.reindex(age_order).dropna(how='all')

result_normalized = result.div(result.max())

result_normalized.plot(kind='bar', stacked=True, figsize=(10,6))

plt.title('Median Compensation and Job Satisfaction by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Normalized Median')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Metric')
plt.tight_layout()
plt.show()
```

### Task 3: Comparing Data Using Stacked Charts


##### 1. Stacked Chart of Preferred Databases by Age Group




Visualize the top databases that respondents from different age groups wish to learn. Create a stacked chart to show the proportion of each database in each age group.


```python
data = df[['Age', 'DatabaseWantToWorkWith']].dropna().copy()

data['DatabaseWantToWorkWith'] = data['DatabaseWantToWorkWith'].str.split(';')
data = data.explode('DatabaseWantToWorkWith').reset_index(drop=True)
data['DatabaseWantToWorkWith'] = data['DatabaseWantToWorkWith'].str.strip()

top_databases = data['DatabaseWantToWorkWith'].value_counts().head(5).index

data = data[
    data['DatabaseWantToWorkWith'].isin(top_databases)
].reset_index(drop=True)

result = pd.crosstab(
    data['Age'],
    data['DatabaseWantToWorkWith'],
    normalize='index'
) * 100

age_order = [
    'Under 18 years old',
    '18-24 years old',
    '25-34 years old',
    '35-44 years old',
    '45-54 years old',
    '55-64 years old',
    '65 years or older'
]

result = result.reindex(age_order).dropna(how='all')

result.plot(
    kind='bar',
    stacked=True,
    figsize=(11,6)
)

plt.title('Preferred Databases by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Share (%)')
plt.xticks(rotation=45, ha='right')
plt.legend(
    title='Database',
    bbox_to_anchor=(1.02,1),
    loc='upper left'
)
plt.tight_layout()
plt.show()
```

##### 2. Stacked Chart of Employment Type by Job Satisfaction


Analyze the distribution of employment types within each job satisfaction level using a stacked chart. This will provide insights into how employment types are distributed across various satisfaction ratings.


```python
data = df[['Employment', 'JobSatPoints_6']].dropna().copy()

data['Employment'] = data['Employment'].str.split(';')
data = data.explode('Employment').reset_index(drop=True)
data['Employment'] = data['Employment'].str.strip()

data['JobSatPoints_6'] = pd.to_numeric(
    data['JobSatPoints_6'],
    errors='coerce'
)

data = data.dropna().reset_index(drop=True)

data['SatisfactionLevel'] = pd.cut(
    data['JobSatPoints_6'],
    bins=[-1, 20, 40, 60, 80, 100],
    labels=['0-20', '21-40', '41-60', '61-80', '81-100']
)

top_employment = data['Employment'].value_counts().head(6).index

data = data[
    data['Employment'].isin(top_employment)
].reset_index(drop=True)

result = pd.crosstab(
    data['SatisfactionLevel'],
    data['Employment'],
    normalize='index'
) * 100

result.plot(
    kind='bar',
    stacked=True,
    figsize=(11,6)
)

plt.title('Employment Type by Job Satisfaction')
plt.xlabel('Job Satisfaction')
plt.ylabel('Share (%)')
plt.xticks(rotation=0)
plt.legend(
    title='Employment Type',
    bbox_to_anchor=(1.02,1),
    loc='upper left'
)
plt.tight_layout()
plt.show()
```

### Task 4: Exploring Technology Preferences Using Stacked Charts


##### 1. Stacked Chart for Preferred Programming Languages by Age Group


Analyze how programming language preferences (`LanguageAdmired`) vary across age groups.


```python
data = df[['Age', 'LanguageAdmired']].dropna().copy()

data['LanguageAdmired'] = data['LanguageAdmired'].str.split(';')
data = data.explode('LanguageAdmired').reset_index(drop=True)
data['LanguageAdmired'] = data['LanguageAdmired'].str.strip()

top_languages = data['LanguageAdmired'].value_counts().head(5).index

data = data[
    data['LanguageAdmired'].isin(top_languages)
].reset_index(drop=True)

result = pd.crosstab(
    data['Age'],
    data['LanguageAdmired'],
    normalize='index'
) * 100

age_order = [
    'Under 18 years old',
    '18-24 years old',
    '25-34 years old',
    '35-44 years old',
    '45-54 years old',
    '55-64 years old',
    '65 years or older'
]

result = result.reindex(age_order).dropna(how='all')

result.plot(
    kind='bar',
    stacked=True,
    figsize=(11,6)
)

plt.title('Preferred Programming Languages by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Share (%)')
plt.xticks(rotation=45, ha='right')
plt.legend(
    title='Programming Language',
    bbox_to_anchor=(1.02,1),
    loc='upper left'
)
plt.tight_layout()
plt.show()
```

##### 2. Stacked Chart for Technology Adoption by Employment Type


Explore how admired platforms (`PlatformAdmired`) differ across employment types (e.g., full-time, freelance)


```python
data = df[['Employment', 'PlatformAdmired']].dropna().copy()

data['Employment'] = data['Employment'].str.split(';')
data = data.explode('Employment').reset_index(drop=True)
data['Employment'] = data['Employment'].str.strip()

data['PlatformAdmired'] = data['PlatformAdmired'].str.split(';')
data = data.explode('PlatformAdmired').reset_index(drop=True)
data['PlatformAdmired'] = data['PlatformAdmired'].str.strip()

top_employment = data['Employment'].value_counts().head(6).index
top_platforms = data['PlatformAdmired'].value_counts().head(5).index

data = data[
    data['Employment'].isin(top_employment) &
    data['PlatformAdmired'].isin(top_platforms)
].reset_index(drop=True)

result = pd.crosstab(
    data['Employment'],
    data['PlatformAdmired'],
    normalize='index'
) * 100

result.plot(
    kind='bar',
    stacked=True,
    figsize=(11,6)
)

plt.title('Admired Platforms by Employment Type')
plt.xlabel('Employment Type')
plt.ylabel('Share (%)')
plt.xticks(rotation=45, ha='right')
plt.legend(
    title='Platform',
    bbox_to_anchor=(1.02,1),
    loc='upper left'
)
plt.tight_layout()
plt.show()
```
